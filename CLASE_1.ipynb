# Palabras Reservadas
Las palabras reservadas (keywords) corresponden a los nombres de las declaraciones que el interprete de Python incluye por defecto, No se deben utilizar para asignar nombres a otros objetos.

**Ejemplo:**

El siguiente comando desplegara la lista de palabras reservadas:

```python
saludo ="hola"
print(saludo)
```

```python
numero = 7
print(numero)
numero2 =5+1
print(numero2)
numero2 =numero+1
print(numero2)
numero3=2
print(numero3)
numero4 = numero3+1
print(numero4)
numero5 = 45.9
print(numero5)
Aprendio =True
Reprobo =False
print(Aprendio)
print(Reprobo)
```
